# Buckled honeycomb lattice: silicene-like geometries
So far the honeycomb lattices we built were flat: both sublattices sit at $z=0$, and the structure has a mirror symmetry $z\to -z$. Materials like silicene, germanene, and stanene instead have a *buckled* honeycomb lattice, where the two sublattices are displaced to $z=\pm\delta$. Breaking that mirror symmetry is what allows a perpendicular electric field to couple linearly to the electrons and (together with spin-orbit coupling) open or close a topological gap — the mechanism explored in the next notebook.
### Import the necessary libraries

```python
# Add the root path of the pyqula library
import os ; import sys
sys.path.append(os.getcwd()+"/../../src")

import numpy as np
import matplotlib.pyplot as plt
from pyqula import geometry
```

### Build the buckled geometry and its first-neighbor Hamiltonian
`geometry.buckled_honeycomb_lattice()` returns a honeycomb lattice with the same in-plane connectivity as before, but with the two sublattices displaced out of plane by $\pm\delta$.
$$ H = \sum_{\langle ij \rangle} c^\dagger_i c_j $$

```python
g = geometry.buckled_honeycomb_lattice() # create the buckled geometry
h = g.get_hamiltonian() # create hamiltonian of the system
print("sublattice z positions:", g.z)
```

### Visualize the buckling: side view of a supercell

```python
gs = g.get_supercell(6)
plt.scatter(gs.x, gs.z, s=40)
plt.xlabel("x") ; plt.ylabel("z")
plt.gca().set_aspect("equal")
plt.show()
```

### Diagonalize the Bloch Hamiltonian
The following function will obtain the Bloch eigenvalues $\epsilon_k$ defined as
$$ H |\Psi_k \rangle = \epsilon_k |\Psi_k\rangle $$

```python
(k,e) = h.get_bands() # get the bandstructure
```

### Plot the Bloch band structure in the path

```python
plt.scatter(k/max(k),e,c="red") # do a scatter plot
plt.ylabel("Energy") # set the y label
plt.xticks([0.,1/3.,0.5,2./3.,1.],["$\Gamma$","$K$","$M$","$K'$","$\Gamma$"])
plt.show() # show the plot
```

The Dirac cones are identical to the flat honeycomb lattice: with only first-neighbor hopping, the bands depend on the in-plane connectivity, not on the out-of-plane embedding. Buckling only becomes electronically relevant once a term that couples to $z$ is added, such as spin-orbit coupling or a perpendicular electric field — see the next notebook.
