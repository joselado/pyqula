# Band structures with state-resolved expectation values
`get_bands(operator=...)` diagonalizes $H(\mathbf k)$ and, for every eigenstate, additionally
returns $\langle\psi_k|\hat O|\psi_k\rangle$ for a chosen operator $\hat O$ — a string name (see
`pyqula.operatorlist` for the full list: `"sz"`, `"valley"`, `"sublattice"`, `"electron"`,
`"surface"`, ...), a custom function, or a matrix. Here we resolve the valley polarization of a
zigzag graphene ribbon.
### Import the necessary libraries

```python
# Add the root path of the pyqula library
import os ; import sys
sys.path.append(os.getcwd()+"/../../../src")

import numpy as np
import matplotlib.pyplot as plt
from pyqula import geometry
from pyqula import islands
```

```python
g = geometry.honeycomb_zigzag_ribbon(10)
h = g.get_hamiltonian(has_spin=False)
(k,e,c) = h.get_bands(operator="valley") # c = valley expectation value per state

plt.scatter(k,e,c=c,cmap="rainbow",s=6)
plt.colorbar(label="valley")
plt.xlabel("k-path") ; plt.ylabel("Energy")
plt.title("Valley-resolved bands of a zigzag ribbon")
plt.show()
```
